# Compute the effective spin Hamiltonian of a system
Compute the effective Hamiltonian in the spin basis

```python
from tranci.atom import get_atom
import numpy as np

ne = 2
# generate an atom class for the system you want to compute
Atom = get_atom(ne=ne) 
# total Hamiltonian
V = Atom.Operator["Coulomb"]
CF = Atom.Operator["x2"] #+ Atom.Operator["x4"]
#CF = Atom.Operator["x4"] + Atom.Operator["y4"] + Atom.Operator["z4"]
LS = Atom.Operator["ls"]
H0 = 4*V + 0.1*CF  # original Hamiltonian 
H = H0 + 0.04*LS # original Hamiltonian 
# degeneracy without magnetic field
M = Atom.get_manifolds(H)
M0 = Atom.get_manifolds(H0)

deg = Atom.get_manifolds(H0).get_gs_multiplicity()
print(deg)
deg = int(np.round(deg,1)) # degeneracy (as integer)
```

```python
from tranci import effectivehamiltonian
#deg = 4
print("Hamiltonian projected on the ",deg," lowest states")
text = effectivehamiltonian.effective_spin_hamiltonian(M,H=H,n=deg)
```
